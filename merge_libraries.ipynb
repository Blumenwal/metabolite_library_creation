```python
import sqlite3
```

### Step 1 - load the individual sqlite files from MoNA, GNPS and MassBank and merge them together ###

```python
### positive mode
con = sqlite3.connect("../combined_pos.sqlite")
con.execute("""CREATE TABLE IF NOT EXISTS spectra(
compound_id TEXT, name TEXT, inchikey TEXT, formula TEXT,
adduct TEXT, precursor_mz REAL, polarity TEXT,
collision_energy TEXT, mz TEXT, intensity TEXT, library_source TEXT)""")

con.execute("""ATTACH DATABASE 'mona_pos.sqlite' AS mona""")
con.execute("""ATTACH DATABASE 'gnps_pos.sqlite' AS gnps""")
con.execute("""ATTACH DATABASE 'massbank_pos.sqlite' AS massbank""")


con.execute("""INSERT INTO spectra SELECT * FROM mona.spectra""")
con.execute("""INSERT INTO spectra SELECT * FROM gnps.spectra""")
con.execute("""INSERT INTO spectra SELECT * FROM massbank.spectra""")

con.commit()

con.execute("""DETACH mona"""); con.execute("""DETACH gnps"""); con.execute("""DETACH massbank""")
con.close()
```

```python
### negative mode
con = sqlite3.connect("../combined_neg.sqlite")
con.execute("""CREATE TABLE IF NOT EXISTS spectra(
compound_id TEXT, name TEXT, inchikey TEXT, formula TEXT,
adduct TEXT, precursor_mz REAL, polarity TEXT,
collision_energy TEXT, mz TEXT, intensity TEXT, library_source TEXT)""")

con.execute("""ATTACH DATABASE 'mona_neg.sqlite' AS mona""")
con.execute("""ATTACH DATABASE 'gnps_neg.sqlite' AS gnps""")
con.execute("""ATTACH DATABASE 'massbank_neg.sqlite' AS massbank""")


con.execute("""INSERT INTO spectra SELECT * FROM mona.spectra""")
con.execute("""INSERT INTO spectra SELECT * FROM gnps.spectra""")
con.execute("""INSERT INTO spectra SELECT * FROM massbank.spectra""")


con.commit()

con.execute("""DETACH mona"""); con.execute("""DETACH gnps"""); con.execute("""DETACH massbank""")
con.close()
```
