```python
import sqlite3
```

### Step 1 - load the individual sqlite files from MoNA, GNPS and MassBank and merge them together ###

```python
### positive mode
con = sqlite3.connect("/home/walesch/PycharmProjects/metabolite_library_creation/combined_pos_instruments.sqlite")
con.execute("""CREATE TABLE IF NOT EXISTS spectra (
compound_id TEXT, name TEXT, inchikey TEXT, formula TEXT,
adduct TEXT, precursor_mz REAL, polarity TEXT,
collision_energy TEXT, mz TEXT, intensity TEXT, library_source TEXT,
instrument_type TEXT, instrument TEXT, exactmass REAL,
inchi TEXT, splash TEXT)""")

con.execute("""ATTACH DATABASE '/home/walesch/PycharmProjects/metabolite_library_creation/mona_pos.sqlite' AS mona""")
con.execute("""ATTACH DATABASE '/home/walesch/PycharmProjects/metabolite_library_creation/gnps_pos.sqlite' AS gnps""")
con.execute("""ATTACH DATABASE '/home/walesch/PycharmProjects/metabolite_library_creation/massbank_pos.sqlite' AS massbank""")


con.execute("""INSERT INTO spectra SELECT * FROM mona.spectra""")
con.execute("""INSERT INTO spectra SELECT * FROM gnps.spectra""")
con.execute("""INSERT INTO spectra SELECT * FROM massbank.spectra""")

con.commit()

con.execute("""DETACH mona"""); con.execute("""DETACH gnps"""); con.execute("""DETACH massbank""")
con.close()
```
